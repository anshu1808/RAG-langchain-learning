```python
from langchain_google_genai import GoogleGenerativeAI
```

```python
import os
```

```python
llm = GoogleGenerativeAI(model="gemini-3.8-flash",
                         temperature=0.7,
                         max_output_tokens=256,
                         google_api_key=os.getenv("GOOGLE_API_KEY"))
```

```python
prompts = [{"role": "user", "content": "How are you?"},
           {"role": "ai", "content": "I'm doing well, thank you! How can I assist you today?"},
           {"role": "user", "content": "Hii, I am anshu from KENEXAI. I want to learn about LLMs and their applications."}]
res = llm.invoke(prompts)
```

########## MANAGE HISTORY

```python
history = []
while True:
    user_input = input("User: ")
    if user_input.lower() in ["exit", "quit"]:
        break

    history.append({"role": "user", "content": user_input})
    print(f"User: {user_input}")

    
    res = llm.invoke(history)
    print(f"AI: {res}")
```
