```python
import os
from dotenv import load_dotenv, find_dotenv
from google import genai
_ = load_dotenv(find_dotenv())  # read local .env file
```

```python
client = genai.Client(api_key=os.environ.get("GENAI_API_KEY"))

response = client.chat(
    model="gemini-3.5-flash-lite",
    contents="Explain Kubernetes in one paragraph."
)
```

```python
def get_completion(prompt, model="gemini-3.5-flash-lite"):
    messages = [
        {"role": "user", "content": prompt}
    ]
    response = genai.client(
        model=model,
        messages=messages,
        temperature=0,
    )
    return response.choices[0].message[ "content" ]
```

```python
def get_response_from_messages(messages, model="gemini-3.5-flash-lite"):
    response = genai.client(
        model=model,
        messages=messages,
        temperature=0,
    )
    return response.choices[0].message["content"]
```

```python
messages = [
    {"role": "system", "content": "You are a assistant like a data engineer."},
    {"role": "user", "content": "What is Kubernetes?"},
    {"role": "assistant", "content": "what concept we can use for data engineer reated to AI?"},
    {"role": "user", "content": "Can you explain it in one paragraph?"}
]
```

```python
response = get_response_from_messages(messages)
print(response)
```
