```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["SERPER_API_KEY"] = os.getenv("SERPER_API_KEY")
os.environ["groq_api_key"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent

search = GoogleSerperAPIWrapper()
llm = ChatGroq(model="openai/gpt-oss-120b", api_key=os.getenv("GROQ_API_KEY"))
```

```python
result = search.run("What is the capital of France?")
```

```python
result
```

```python
raw_result = search.results("langchain updates 2026")
raw_result
```

Integrate into a LangChain Agent

```python
agent = create_agent(model = llm, 
                     tools=[search.run],
                     system_prompt="You are a helpful assistant that can answer questions using Google Search.",
                     )
```

```python
question = "What is the latest news about LangChain in 2026?"
```

```python
res = agent.invoke({"message": [{"role": "user", "content": question}]})
```

```python
print(res['messages'][-1])
```
