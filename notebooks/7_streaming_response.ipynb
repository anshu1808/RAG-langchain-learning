```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model="gpt-4o", streaming=True, verbose=True, openai_api_key=os.environ.get("OPENAI_API_KEY"))
```

```python
question = "who is the president of the India?"
```

```python
response = llm.invoke(question)
response.content
```

```python
from langchain_groq import ChatGroq
import os
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b", api_key=os.environ.get("GROQ_API_KEY"), streaming=True, verbose=True)
```

```python
question = "can you teach me the learning steps of AI?"
response = llm.stream(question)
```

```python
for chunk in response:
    print(chunk.content, end="")
```
