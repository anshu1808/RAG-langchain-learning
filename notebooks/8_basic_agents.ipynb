```python
from langchain.agents import create_agent
from langchain.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together.
    
       Args:
           a (int): The first number.
           b (int): The second number.
    """
    return a + b
    
```

```python
@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers together.
    
       Args:
           a (int): The first number.
           b (int): The second number.
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 10})
```

```python
multiply_numbers.invoke({"a": 5, "b": 10})
```

```python
from langchain_groq import ChatGroq
from langchain_ollama import ChatOllama
from langchain_google_genai import ChatGoogleGenerativeAI
import os
```

```python
#llm = ChatOllama(model="smollm2:360m")
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash",
                         temperature=0.5,
                         max_output_tokens=256,
                         google_api_key=os.getenv("GOOGLE_API_KEY"))
```

```python
agent = create_agent(model=llm, tools = [add_numbers, multiply_numbers],
                     system_prompt="You are a helpful assistant that can do mathematical calculations. Use the tools provided to perform calculations."
                     )
```

```python
res = agent.invoke({"messages":[{"role": "user", "content": "do mathematical calculations for 33 and 10?"}]})
```
