```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
import getpass

if not os.getenv("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your google API key: ")
```

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2-preview",
    output_dimensionality=768,  # Optional: reduce size for storage optimization
)
```

```python
vector = embeddings.embed_documents(documents)
```

```python
len(vector[0])
```

##store vector embedding into vector db

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db",  # Where to save data locally, remove if not necessary
)
```

```python
query = "Machine learnings"
result = vector_store.similarity_search(query = query)
result
```
