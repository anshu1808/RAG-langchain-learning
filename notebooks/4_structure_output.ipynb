```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.7
)
```

```python
print(type(llm))
```

```python
text = "HI, I am Anshu from KENEXAI. email is anshu.accent@gmail.com. i am a software engineer. i am 29 years old"
```

```python
from pydantic import BaseModel, Field

class UserInfo(BaseModel):
    name: str = Field(..., description="The name of the user")
    email: str = Field(..., description="The email address of the user")
    profession: str = Field(..., description="The profession of the user")
    age: int = Field(..., description="The age of the user")

structured_output = llm.with_structured_output(UserInfo)
```

```python
res = structured_output.invoke("can you extract the email address from this text: " + text)
```

```python
res
```

```python
res.name
```

```python
res.model_dump()
```

```python
from pydantic import BaseModel, Field, field_validator

class profile(BaseModel):
    name: str = Field(..., description="The name of the user")
    email: str = Field(..., description="The email address of the user")
    profession: str = Field(..., description="The profession of the user")
    age: int = Field(..., description="The age of the user")

    @field_validator("email")
    def validate_email(cls, value):
        if "@" not in value:
            raise ValueError("Invalid email address")
        return value

structured_output = llm.with_structured_output(profile)
```

```python
structured_output.invoke("can you verify email is valid?")
```

```python

text = """
My name is Anshu.
My email is anshu.accent@gmail.com.
I am a software engineer.
I am 29 years old.
"""

result = structured_output.invoke(text)
print(result)
```

```python
class Movies(BaseModel):
    title: str = Field(..., description="The title of the movie")
    director: str = Field(..., description="The director of the movie")
    release_year: int = Field(..., description="The release year of the movie")
    genre: str = Field(..., description="The genre of the movie")

class AllMovies(BaseModel):
    movies: list[Movies] = Field(..., description="A list of movies")

movie_output = llm.with_structured_output(AllMovies)
```

```python
movie_output.invoke("suggest four thriller movie")
```
